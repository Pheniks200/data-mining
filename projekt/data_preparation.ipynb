```python
#tabela kontyngencji (zmeinne jakosciowe)
from scipy.stats import chi2_contingency
from itertools import combinations

df = bank_marketing

kolumny_jakosciowe = df.select_dtypes(include=['object', 'category']).columns

wyniki_testow = []

for zm1, zm2 in combinations(kolumny_jakosciowe, 2):
    
    tabela = pd.crosstab(df[zm1], df[zm2])
    
    if tabela.size == 0:
        continue
        
    chi2, p_value, dof, expected = chi2_contingency(tabela)
    
    status_zaleznosci = "ZALEŻNE" if p_value < 0.05 else "NIEZALEŻNE"
    
    wyniki_testow.append({
        'Zmienna 1': zm1,
        'Zmienna 2': zm2,
        'Statystyka Chi2': round(chi2, 4),
        'p-value': p_value,
        'Czy zależne? (alpha=0.05)': status_zaleznosci
    })

raport_chi2 = pd.DataFrame(wyniki_testow)

raport_chi2 = raport_chi2.sort_values(by='p-value')

print(raport_chi2.to_string(index=False))

raport_chi2.to_excel('raport_testow_chi_kwadrat.xlsx', index=False)
```

```python
# Identyfikacja zmiennych z brakami danych (nan)
bank_marketing.isnull().sum()
```

```python
# Identyfikacja duplikatów
bank_marketing.duplicated().sum()
```

```python
# Usuwanie duplikatów
bank_marketing = bank_marketing.drop_duplicates()
```

```python
# Identyfikacja zmiennych z brakami danych (unknown)
(bank_marketing == 'unknown').sum()
```

```python
# Zamiana "unknown" na nan
bank_marketing = bank_marketing.replace('unknown', np.nan)
```

```python
# Poprawa wartości dla kolumny pdays
# Opis mówi: 999 means client was not previously contacted
# To nie jest normalna wartość liczby dni, tylko specjalny kod 
# Należy poprawić na nan
bank_marketing["pdays"] = bank_marketing["pdays"].replace(999, np.nan)
```

```python
# duration = czas ostatniej rozmowy
# Wartość zmiennej jest znana dopiero po zakończeniu rozmowy. W modelu predykcyjnym stanowi ona wyciek danych (leakage), z tego powodu należy ją usunąć
bank_marketing = bank_marketing.drop(columns=['duration'])
```

```python
# Proporcje targetu
bank_marketing['y'].value_counts(normalize=True)  # Widoczna nierównowaga - dużo więcej no niż yes
```

```python
# Podział na train / validation / test

# Podział train (70%) oraz temp (30%)
train_df, temp_df = train_test_split(bank_marketing, test_size=0.30, random_state=1, stratify=bank_marketing['y'])

# Podział temp na validation i test
valid_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=1, stratify=temp_df['y'])

# Dodanie informacji o zbiorze
train_df['split'] = 'TRAIN'
valid_df['split'] = 'VALID'
test_df['split'] = 'TEST'

# Undersampling klasy dominującej tylko dla zbioru treningowego

# Podział klas
train_yes = train_df[train_df['y'] == 'yes']
train_no = train_df[train_df['y'] == 'no']

# Undersampling klasy majority
# Ustawiamy proporcję około 2:3
train_no_sample = train_no.sample(n=int(len(train_yes)*1.5), random_state=1)

# Połączenie danych
train_balanced = pd.concat([train_yes, train_no_sample])

# Przemieszanie rekordów
train_balanced = train_balanced.sample(frac=1, random_state=1).reset_index(drop=True)

# Kontrola proporcji klas
train_balanced['y'].value_counts(normalize=True)


# Połączenie wszystkiego w jedną tabelę
final_df = pd.concat([train_balanced, valid_df, test_df]).reset_index(drop=True)

# Zapis do CSV
final_df.to_csv('bank_marketing_split.csv', index=False)
```
